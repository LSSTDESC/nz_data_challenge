# Notebook to run evaluation on all submissions

#### Standard imports

```python
import os
from nz_data_challenge import evaluation
from pathlib import Path
```

#### Paths

```python
submissions = [
    'example'
]
test_only = True
if test_only:
    submit_top_dir = '../submission_test'
    public_dir = '../public'
    results_top_dir = '../results_test'
    truth_dir = '../public'
    test_suffix = 'ddf_00'
else:
    submit_top_dir = '../submission'
    public_dir = '../public'
    results_top_dir = '../results'
    truth_dir = '../reserved'
    test_suffix = 'wfd'

template_jinja_file = '../docs_results/_templates/entry_summary.rst.j2'

try:
    os.makedirs(submit_dir)
except:
    pass
try:
    os.makedirs(results_dir)
except:
    pass
```

#### Call the evaluation function

```python
evaluation.evaluate_all_submissions(
    submissions,
    submit_top_dir,
    public_dir,
    truth_dir,
    results_top_dir,
    template_jinja_file,
    suffix=test_suffix,
)
```
