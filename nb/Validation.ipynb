# Notebook to make validation plots

This makes the validation plots used in the instructions documentaion

#### Standard imports

```python
import tables_io, qp
import numpy as np
import matplotlib.pyplot as plt

from rail.core.data import TableHandle
from rail.utils import catalog_utils
from nz_data_challenge import validation, utils
```

#### Paths for nz challenge related files

```python
public_dir = '../public'
figure_dir = '../docs/figures'
```

#### Loop over stuff and make the validation plots

```python
# tasksets = ['taskset_1', 'taskset_2']
# sims = ['cardinal', 'flagship']
# scenarios = ['1yr', '4yr']

fig = plt.figure(figsize=(10, 10))
axes = fig.subplots(2, 4)
idx = 0
for taskset in utils.TASKSETS:
    for sim in utils.SIMS:
        for scenario in utils.SCENARIOS:
            wfd, ddf_list = validation.get_data(public_dir, taskset, sim, scenario)
            fig_footprint = validation.draw_footprint(wfd, ddf_list)
            fig_rubin_mags = validation.draw_rubin_mags(wfd, ddf_list[0])
            fig_roman_mags = validation.draw_roman_mags(wfd, ddf_list[0])
            fig_footprint.savefig(f"{figure_dir}/nz_challenge_{taskset}_{sim}_{scenario}_footprint.png")
            fig_rubin_mags.savefig(f"{figure_dir}/nz_challenge_{taskset}_{sim}_{scenario}_rubin_mags.png")
            fig_roman_mags.savefig(f"{figure_dir}/nz_challenge_{taskset}_{sim}_{scenario}_roman_mags.png")            
```
